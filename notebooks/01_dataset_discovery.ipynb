```python
from tdc.multi_pred import DDI
import pandas as pd
from pathlib import Path

print("Imports successful!")
```

```python
DATA_PATH = Path("../data/raw/tdc")
DATA_PATH.mkdir(parents=True, exist_ok=True)

print("Data will be stored in:", DATA_PATH.resolve())
```

```python
drugbank_data = DDI(
    name="DrugBank",
    path=str(DATA_PATH)
)

print("DrugBank DDI dataset object created successfully!")
```

```python
drugbank_df = drugbank_data.get_data()

print("Dataset loaded successfully!")
print("Shape:", drugbank_df.shape)
```

```python
drugbank_df.head()
```

```python
print("Columns:")
print(drugbank_df.columns.tolist())

print("\nData types:")
print(drugbank_df.dtypes)

print("\nMissing values:")
print(drugbank_df.isnull().sum())
```

```python
drugbank_df["Y"].value_counts(dropna=False)
```

```python
drugbank_df["Y"].unique()
```

```python
exact_duplicates = drugbank_df.duplicated().sum()
print("Exact duplicate rows:", exact_duplicates)
```

```python
pair_duplicates = drugbank_df.duplicated(
    subset=["Drug1_ID", "Drug2_ID"],
    keep=False
).sum()

print("Rows belonging to repeated directed pairs:", pair_duplicates)
```

```python
repeated_pairs = drugbank_df[
    drugbank_df.duplicated(
        subset=["Drug1_ID", "Drug2_ID"],
        keep=False
    )
].sort_values(
    ["Drug1_ID", "Drug2_ID", "Y"]
)

repeated_pairs.head(20)
```

```python
pair_summary = (
    repeated_pairs
    .groupby(["Drug1_ID", "Drug2_ID"])
    .agg(
        interaction_types=("Y", list),
        num_interaction_types=("Y", "nunique"),
        total_records=("Y", "size")
    )
    .reset_index()
)

pair_summary.head(20)
```

```python
duplicate_relationships = drugbank_df.duplicated(
    subset=["Drug1_ID", "Drug2_ID", "Y"]
).sum()

print("Duplicate drug-pair + interaction-type records:", duplicate_relationships)
```

```python
pair_set = set(
    zip(
        drugbank_df["Drug1_ID"],
        drugbank_df["Drug2_ID"]
    )
)

reverse_pair_count = sum(
    1
    for drug1, drug2 in pair_set
    if (drug2, drug1) in pair_set
)

print("Directed pairs having a reverse counterpart:", reverse_pair_count)
print("Total unique directed pairs:", len(pair_set))
```

```python
unique_drugs = set(drugbank_df["Drug1_ID"]) | set(drugbank_df["Drug2_ID"])

print("Unique drugs:", len(unique_drugs))
```

```python
drug1_map = (
    drugbank_df.groupby("Drug1_ID")["Drug1"]
    .nunique()
)

drug2_map = (
    drugbank_df.groupby("Drug2_ID")["Drug2"]
    .nunique()
)

print("Drug1 IDs with multiple structures:", (drug1_map > 1).sum())
print("Drug2 IDs with multiple structures:", (drug2_map > 1).sum())
```

```python
from pathlib import Path

twosides_path = Path("../data/raw/twosides")
twosides_path.mkdir(parents=True, exist_ok=True)

print("Download path:", twosides_path.resolve())
```

```python
from tdc.multi_pred import DDI

twosides_data = DDI(
    name="TWOSIDES",
    path="../data/raw/twosides"
)

print("TWOSIDES dataset object created successfully!")
```

```python
twosides_df = twosides_data.get_data()

print("Dataset loaded successfully!")
print("Shape:", twosides_df.shape)
```

```python
print("Columns:")
print(twosides_df.columns.tolist())

print("\nData types:")
print(twosides_df.dtypes)

print("\nMissing values:")
print(twosides_df.isnull().sum())

print("\nFirst 5 rows:")
display(twosides_df.head())
```

```python
print("\nUnique Y values:", twosides_df["Y"].nunique())

print("\nTop interaction/side-effect labels:")
print(twosides_df["Y"].value_counts().head(20))
```

```python
print("\nUnique Y labels:")
print(sorted(twosides_df["Y"].unique())[:20])
```

```python
print(
    "Exact duplicate rows:",
    twosides_df.duplicated().sum()
)
```

```python
print(
    "Duplicate Drug1 + Drug2 + Y relationships:",
    twosides_df.duplicated(
        subset=["Drug1_ID", "Drug2_ID", "Y"]
    ).sum()
)
```

```python
repeated_pair_rows = twosides_df.duplicated(
    subset=["Drug1_ID", "Drug2_ID"],
    keep=False
).sum()

print(
    "Rows belonging to repeated directed drug pairs:",
    repeated_pair_rows
)
```

```python
unique_directed_pairs = twosides_df[
    ["Drug1_ID", "Drug2_ID"]
].drop_duplicates()

print(
    "Unique directed drug pairs:",
    len(unique_directed_pairs)
)
```

```python
events_per_pair = (
    twosides_df
    .groupby(["Drug1_ID", "Drug2_ID"])
    .size()
)

print("Average adverse events per pair:", events_per_pair.mean())
print("Minimum:", events_per_pair.min())
print("Maximum:", events_per_pair.max())
```

```python
events_per_pair.sort_values(
    ascending=False
).head(10)
```

```python
unique_drugs = set(twosides_df["Drug1_ID"]) | set(twosides_df["Drug2_ID"])

print("Unique drugs:", len(unique_drugs))
```

```python
drug1_structure_check = (
    twosides_df
    .groupby("Drug1_ID")["Drug1"]
    .nunique()
)

drug2_structure_check = (
    twosides_df
    .groupby("Drug2_ID")["Drug2"]
    .nunique()
)

print(
    "Drug1 IDs with multiple structures:",
    (drug1_structure_check > 1).sum()
)

print(
    "Drug2 IDs with multiple structures:",
    (drug2_structure_check > 1).sum()
)
```

```python
pairs = twosides_df[
    ["Drug1_ID", "Drug2_ID"]
].drop_duplicates()

pair_set = set(
    zip(
        pairs["Drug1_ID"],
        pairs["Drug2_ID"]
    )
)

reverse_pair_count = sum(
    (drug2, drug1) in pair_set
    for drug1, drug2 in pair_set
)

print("Directed pairs having a reverse counterpart:", reverse_pair_count)
print("Total unique directed pairs:", len(pair_set))
```

```python
import pandas as pd

raw_twosides = pd.read_csv("../data/raw/twosides/twosides.csv")

print("Shape:", raw_twosides.shape)
print("\nColumns:", raw_twosides.columns.tolist())

display(raw_twosides.head())
```

```python
print("Raw TWOSIDES shape:", raw_twosides.shape)

print("\nColumns:")
print(raw_twosides.columns.tolist())

print("\nMissing values:")
print(raw_twosides.isnull().sum())

print("\nUnique side-effect names:")
print(raw_twosides["Side Effect Name"].nunique())

print("\nUnique Y labels:")
print(raw_twosides["Y"].nunique())
```

```python
y_to_name = (
    raw_twosides
    .groupby("Y")["Side Effect Name"]
    .nunique()
)

print(
    "Y labels mapping to more than one side-effect name:",
    (y_to_name > 1).sum()
)
```

```python
name_to_y = (
    raw_twosides
    .groupby("Side Effect Name")["Y"]
    .nunique()
)

print(
    "Side-effect names mapping to more than one Y:",
    (name_to_y > 1).sum()
)
```
